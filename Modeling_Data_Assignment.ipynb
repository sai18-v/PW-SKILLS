## Q1. What is Data Loading in Power BI and why is it considered the first step of analysis?

**Answer:**
Data Loading is the process of importing data from a source into Power BI for analysis. It is considered the first step because the data must be available in Power BI before it can be cleaned, transformed, modeled, and visualized.

---

## Q2. Explain the difference between “Load” and “Transform Data” in Power BI.

**Answer:**

* **Load:** Directly imports the selected data into Power BI without making transformations.
* **Transform Data:** Opens Power Query Editor, where data can be cleaned, filtered, changed, and transformed before loading it into Power BI.

---

## Q3. What is a Fact Table and a Dimension Table? Give examples from the dataset.

**Answer:**

* **Fact Table:** Contains measurable business events or numerical values. In this dataset, the COVID-19 records containing **Confirmed, Deaths, and Recovered** can serve as the fact data.
* **Dimension Table:** Contains descriptive information used to categorize or filter facts. **Country, State, and Date** can serve as dimensions.

---

## Q4. Why is Star Schema preferred over Snowflake Schema in Power BI?

**Answer:**
Star Schema is preferred because it has a simple structure, fewer relationships, better query performance, easier data modeling, and simpler DAX calculations. It consists of a central fact table connected directly to dimension tables.

## Q5. Identify and remove duplicate records based on Date, Country, and State.

**Answer:**

1. Open **Power Query Editor**.
2. Select the **Date**, **Country**, and **State** columns.
3. Go to **Home → Remove Rows → Remove Duplicates**.
4. Power Query keeps one record for each unique combination of Date, Country, and State.

## Q7. Create a new column to calculate Recovery Rate.

**Answer:**

```DAX
Recovery Rate =
DIVIDE(
    'Covid_19'[Recovered],
    'Covid_19'[Confirmed],
    0
) * 100
```

## Q8. Create a summarized table showing total confirmed cases by Country.

**Answer:**

```DAX
CountrySummary =
SUMMARIZE(
    'Covid_19',
    'Covid_19'[Country],
    "Total Confirmed Cases", SUM('Covid_19'[Confirmed])
)
```
