### Question 1

**Question:** Show employees working in either the IT or HR departments.
**Answer:**

```sql
SELECT * FROM Employees
WHERE Department IN ('IT', 'HR');
```

### Question 2

**Question:** Retrieve employees whose department is in Sales, IT, or Finance.
**Answer:**

```sql
SELECT * FROM Employees
WHERE Department IN ('Sales', 'IT', 'Finance');
```

### Question 3

**Question:** Display employees whose salary is between ₹50,000 and ₹70,000.
**Answer:**

```sql
SELECT * FROM Employees
WHERE Salary BETWEEN 50000 AND 70000;
```

### Question 4

**Question:** List employees whose names start with the letter 'A'.
**Answer:**

```sql
SELECT * FROM Employees
WHERE EmpName LIKE 'A%';
```

### Question 5

**Question:** Find employees whose names contain the substring 'an'.
**Answer:**

```sql
SELECT * FROM Employees
WHERE EmpName LIKE '%an%';
```

### Question 6

**Question:** Show employees who are from Delhi or Mumbai and earn more than ₹55,000.
**Answer:**

```sql
SELECT * FROM Employees
WHERE City IN ('Delhi', 'Mumbai')
AND Salary > 55000;
```

### Question 7

**Question:** Display all employees except those from the HR department.
**Answer:**

```sql
SELECT * FROM Employees
WHERE Department <> 'HR';
```

### Question 8

**Question:** Get all employees hired between 2019 and 2022, ordered by HireDate (oldest first).
**Answer:**

```sql
SELECT * FROM Employees
WHERE HireDate BETWEEN '2019-01-01' AND '2022-12-31'
ORDER BY HireDate ASC;
```
