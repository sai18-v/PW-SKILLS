## Q1. Create a Card Visual to show Total Sales.

**Answer:**
Insert a **Card** visual and drag **Sales** into the **Fields/Data** area. Set the aggregation to **Sum**.

## Q3. Create a Line Chart showing the Sales Trend over time (by Month and Year). Is there any noticeable seasonal pattern in sales?

**Answer:**

* Y-axis: **Sales (Sum)**

**Observation:** Yes. Sales show noticeable seasonal variation, with sales generally increasing toward the end of the year.

## Q10. Create a Dual-Axis Line Chart to show Total Sales and Profit Over Time (by Month).

**Answer:**

* Visual: **Line and clustered column chart** (or Line and stacked column chart)
* X-axis: **Order Date (Month and Year)**
* Column Y-axis: **Sales (Sum)**
* Line Y-axis: **Profit (Sum)**

This allows **Sales and Profit** to be compared over time using separate axes.
