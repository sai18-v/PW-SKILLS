**Answer:**
**Incremental Load**

### b) Justify your choice.

**Answer:**
Since the dataset represents **daily sales data**, incremental loading is more appropriate. It loads only new or changed records instead of reloading the entire dataset every day.

### a) What incorrect results might appear in Total Sales KPI?
