### Q1. Explain the difference between a Tableau Dashboard and a Tableau Story Point. Give two healthcare-related use cases.

**Answer:**
A **Tableau Dashboard** combines multiple worksheets and visualizations into one interactive view for monitoring and analysis.
A **Story Point** is a single step in a Tableau Story that presents a particular analysis or insight in sequence.

**Healthcare use cases:**

1. **Dashboard:** Monitor patient count, treatment cost, department performance, and outcomes.
2. **Story Point:** Present a step-by-step analysis of patient distribution → department costs → treatment outcomes → doctor performance.

---

### Q2. What are Actions in Tableau? Explain how Filter Action can help hospital management analyze patient data.

**Answer:**
Actions make Tableau visualizations interactive by allowing one visualization to affect another.

A **Filter Action** allows selecting a value in one visualization to filter related visualizations.

**Example:** Selecting the **South Region** in a region chart can automatically filter the dashboard to show only South-region patients, treatment costs, departments, and outcomes.
