### Q1. Define Data Transformation in ETL and explain why it is important.

**Answer:**
Data transformation is the process of converting extracted data into a consistent and suitable format before loading it into the target system. It is important because it improves data quality, consistency, accuracy, and usability.

1. Handling missing values
2. Removing duplicate records
3. Correcting inconsistent data
4. Identifying and handling outliers

### Q3. What is the difference between Normalization and Standardization?

**Answer:**
**Normalization** scales data to a fixed range, usually **0 to 1**.

**Standardization** transforms data so that it has a **mean of 0 and a standard deviation of 1**.

### Q4. A dataset has missing values in the “Age” column. Suggest two techniques to handle this and explain when they should be used.

**Answer:**

1. **Mean/Median Imputation:** Replace missing ages with the mean or median age. Median is preferred when the data contains outliers or is skewed.
2. **Removing Records:** Delete rows with missing ages when only a small number of values are missing and removing them will not significantly affect the dataset.

### Q6. What is One-Hot Encoding? Give an example with the categories: “Red, Blue, Green”.

### Q7. Explain the difference between Data Integration and Data Mapping in ETL.

**Answer:**
**Data Integration** combines data from multiple sources into a unified system.

**Data Mapping** defines how fields or columns from the source correspond to fields or columns in the target system.

### Q8. Explain why Z-score Standardization is preferred over Min-Max Scaling when outliers exist.

**Answer:**
Z-score Standardization is generally preferred because it is less directly affected by extreme minimum and maximum values than Min-Max Scaling. Min-Max Scaling uses the minimum and maximum values, so an outlier can compress most normal observations into a small range. Z-score Standardization uses the mean and standard deviation to express values relative to the distribution.
