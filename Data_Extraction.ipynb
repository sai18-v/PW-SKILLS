## Question 1: Describe different types of data sources used in ETL with suitable examples.

**Answer:**
Common data sources used in ETL include:

1. **Relational databases** – MySQL, PostgreSQL, Oracle, SQL Server.
2. **Files** – CSV, Excel, JSON, and XML files.
3. **APIs** – REST APIs and web services that provide data.
4. **Cloud storage** – Amazon S3, Google Cloud Storage, and Azure Blob Storage.
5. **NoSQL databases** – MongoDB, Cassandra, and Redis.

## Question 2: What is data extraction? Explain its role in the ETL pipeline.

**Answer:**
Data extraction is the process of collecting data from one or more sources such as databases, files, APIs, or cloud systems.

It is the **first step of the ETL pipeline**. The extracted data is then transformed into the required format and loaded into the target system such as a data warehouse.

**ETL:**
**Extract → Transform → Load**

## Question 4: Explain the steps involved in extracting data from a relational database.

**Answer:**

1. Identify the required database and tables.
2. Establish a connection using database credentials.
3. Write SQL queries to select the required data.
4. Execute the queries.
5. Retrieve the result set.
6. Validate the extracted data.
7. Store or pass the data to the transformation stage.

## Question 5: Explain three common challenges faced during data extraction.

**Answer:**

1. **Large data volume** – Extracting large datasets can consume significant memory and processing time.
2. **Data quality issues** – Missing, duplicate, or inconsistent data may exist in the source.
3. **Connectivity and access issues** – Network failures, authentication problems, or insufficient permissions can interrupt extraction.

## Question 6: What are APIs? Explain how APIs help in real-time data extraction.

**Answer:**
APIs (Application Programming Interfaces) allow different software applications to communicate and exchange data.

In real-time data extraction, an ETL system can call an API and retrieve the latest data directly from the source. This is useful for applications such as live weather information, stock prices, payment systems, and social media data.

## Question 7: Why are databases preferred for enterprise-level data extraction?

**Answer:**
Databases are preferred because they can efficiently handle large volumes of structured data and support SQL queries, indexing, transactions, security, concurrency, and reliable data access. They also provide better performance and scalability for enterprise-level ETL processes.

## Question 8: What steps should an ETL developer take when extracting data from large CSV files (1GB+)?

**Answer:**

1. Use **chunking/batch processing** instead of loading the entire file into memory.
2. Read only the required columns.
3. Process and transform each chunk separately.
4. Validate data during extraction.
5. Handle missing, duplicate, and invalid records.
6. Use appropriate data types to reduce memory usage.
7. Store the processed data in batches.
8. Monitor memory usage, processing time, and errors.
