## 1. What is the purpose of a PivotTable in Excel?

**Answer:**
A PivotTable is used to summarize, analyze, and organize large amounts of data quickly. It helps users calculate totals, averages, counts, and compare data across categories without changing the original dataset.

---

## 2. Create a PivotTable to show Total Sales by Region.

**Answer:**

1. Select the dataset.
2. Go to **Insert → PivotTable**.
3. Place **Region** in the **Rows** area.
4. Place **Sales** in the **Values** area.
5. Ensure Values are summarized by **Sum**.

**Output Example:**

| Region | Total Sales |
| ------ | ----------- |
| East   | ₹5,000      |
| West   | ₹4,500      |
| North  | ₹6,200      |
| South  | ₹5,800      |

---

## 3. Apply Conditional Formatting to highlight sales greater than 400.

**Answer:**

1. Select the Sales column.
2. Go to **Home → Conditional Formatting → Highlight Cells Rules → Greater Than**.
3. Enter **400**.
4. Choose a highlight color.
5. Click **OK**.

Result: All sales values above 400 will be highlighted.

## 4. Difference between Sorting and Filtering.

**Answer:**

| Sorting                                 | Filtering                              |
| --------------------------------------- | -------------------------------------- |
| Rearranges data order                   | Shows only matching records            |
| Example: Sort Sales Largest to Smallest | Example: Show only East Region records |

**Example:**

* Sorting: ₹800, ₹700, ₹500, ₹300
* Filtering: Display only rows where Region = East

## 5. Create a PivotChart showing Monthly Sales Trend.

**Answer:**

1. Create a PivotTable.
2. Drag **Date** to Rows.
3. Group Date by **Months**.
4. Drag **Sales** to Values (Sum).
5. Go to **Insert → PivotChart → Line Chart**.

This shows sales performance month by month.

---

## 6. Use a Slicer to filter data by Product.

**Answer:**

1. Select the PivotTable.
2. Go to **PivotTable Analyze → Insert Slicer**.
3. Select **Product**.
4. Click **OK**.
5. Use the slicer buttons (Apples, Oranges, Bananas, Mangoes) to filter data.

## 9. Group dates into Months and calculate Average Sales per Month.

**Answer:**

1. Create PivotTable.
2. Drag **Date** to Rows.
3. Right-click a date → **Group** → Months.
4. Drag **Sales** to Values.
5. Change Value Field Settings from **Sum** to **Average**.

**Output Example:**

| Month | Average Sales |
| ----- | ------------- |
| Jan   | ₹450          |
| Feb   | ₹480          |
| Mar   | ₹520          |
| Apr   | ₹500          |
| May   | ₹550          |
| Jun   | ₹600          |

## 10. Combine Apples and Oranges into a group called "Fruits" and compare sales.

**Answer:**

1. In PivotTable, select **Apples** and **Oranges**.
2. Right-click → **Group**.
3. Rename the group as **Fruits**.
4. Compare sales against Bananas and Mangoes.

**Example Output:**

| Product Group             | Total Sales |
| ------------------------- | ----------- |
| Fruits (Apples + Oranges) | ₹8,500      |
| Bananas                   | ₹3,200      |
| Mangoes                   | ₹4,100      |

This allows easier comparison between grouped and individual products.
