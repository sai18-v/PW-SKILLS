### 1. What is the key difference between publishing a dashboard on Tableau Public versus Tableau Server/Cloud in terms of accessibility and data security?

**Answer:**
**Tableau Public** dashboards are publicly accessible on the internet, and the published data can generally be downloaded by viewers. It is **not suitable for confidential data**.
**Tableau Server/Cloud** provides controlled access through permissions, authentication, and security settings, making it suitable for private organizational data.

---

### 2. When publishing a Tableau dashboard, what happens to a live connection compared to an extract? How does this affect data refresh and performance?

**Answer:**

* **Live connection:** The dashboard queries the original data source when users interact with it. Data is more current, but performance depends on the source and network.
* **Extract:** A snapshot of the data is stored in Tableau's extract format. It generally provides faster performance, but the data must be refreshed to reflect changes in the source.

---

### 3. In Tableau Server or Cloud, which feature allows administrators to control which users can view, interact with, or edit a published dashboard?

**Answer:**
**Permissions** control what users and groups can **view, interact with, download, or edit** published dashboards.

---

### 4. If you want users to view a dashboard but not download the underlying data, which permission setting would you adjust in Tableau Server?

**Answer:**
Disable the **Download Full Data** permission for the relevant users or groups.

### 6. List two best practices when preparing a dashboard for public sharing to ensure data security and privacy.

**Answer:**

1. **Remove or anonymize sensitive/confidential information** before publishing.
2. **Review and restrict data-download capabilities** and ensure only appropriate aggregated data is exposed.

### 7. After publishing a dashboard, users report a “Data not found” error. What is the most common cause related to data source settings?

**Answer:**
The published workbook cannot access the original data source because the **data-source connection/path or credentials are invalid or unavailable** on the Tableau Server/Cloud environment.

---

### 8. When embedding a Tableau dashboard into a secure internal portal, what two additional security measures should be implemented to ensure only authorized users can access the embedded content?

**Answer:**

1. **Authentication** — require users to authenticate through the organization's identity provider/SSO.
2. **User/group-based permissions** — grant dashboard access only to authorized users or groups.
