### Question 1

**Question:** Retrieve all customers who have placed at least one order.
**Answer:**

```sql
SELECT DISTINCT c.*
FROM Customers c
INNER JOIN Orders o
ON c.CustomerID = o.CustomerID;
```

### Question 2

**Question:** Retrieve all customers and their orders, including customers who have not placed any orders.
**Answer:**

```sql
SELECT c.*, o.*
FROM Customers c
LEFT JOIN Orders o
ON c.CustomerID = o.CustomerID;
```

### Question 3

**Question:** Retrieve all orders and their corresponding customers, including orders placed by unknown customers.
**Answer:**

```sql
SELECT o.*, c.*
FROM Orders o
LEFT JOIN Customers c
ON o.CustomerID = c.CustomerID;
```

### Question 4

**Question:** Display all customers and orders, whether matched or not.
**Answer:**

```sql
SELECT c.*, o.*
FROM Customers c
LEFT JOIN Orders o
ON c.CustomerID = o.CustomerID

UNION

SELECT c.*, o.*
FROM Customers c
RIGHT JOIN Orders o
ON c.CustomerID = o.CustomerID;
```

### Question 5

**Question:** Find customers who have not placed any orders.
**Answer:**

```sql
SELECT c.*
FROM Customers c
LEFT JOIN Orders o
ON c.CustomerID = o.CustomerID
WHERE o.CustomerID IS NULL;
```

### Question 6

**Question:** Retrieve customers who made payments but did not place any orders.
**Answer:**

```sql
SELECT DISTINCT c.*
FROM Customers c
INNER JOIN Payments p
ON c.CustomerID = p.CustomerID
LEFT JOIN Orders o
ON c.CustomerID = o.CustomerID
WHERE o.CustomerID IS NULL;
```

### Question 7

**Question:** Generate a list of all possible combinations between Customers and Orders.
**Answer:**

```sql
SELECT c.*, o.*
FROM Customers c
CROSS JOIN Orders o;
```

### Question 8

**Question:** Show all customers along with order and payment amounts in one table.
**Answer:**

```sql
SELECT c.CustomerID,
       c.CustomerName,
       o.Amount AS OrderAmount,
       p.Amount AS PaymentAmount
FROM Customers c
LEFT JOIN Orders o
ON c.CustomerID = o.CustomerID
LEFT JOIN Payments p
ON c.CustomerID = p.CustomerID;
```

### Question 9

**Question:** Retrieve all customers who have both placed orders and made payments.
**Answer:**

```sql
SELECT DISTINCT c.*
FROM Customers c
INNER JOIN Orders o
ON c.CustomerID = o.CustomerID
INNER JOIN Payments p
ON c.CustomerID = p.CustomerID;
```
