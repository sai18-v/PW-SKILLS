## Q1: Understanding Central Tendency (Easy)

**Question:**
A bakery tracks the daily sales of muffins (in dozens) over a week: **[10, 12, 11, 15, 14, 13, 12]**. What is the most representative value of their weekly sales, and why?

**Answer:**

* Mean = (10 + 12 + 11 + 15 + 14 + 13 + 12) ÷ 7 = **12.43**
* The **mean (≈12.43 dozens)** is the most representative because there are no extreme values.

## Q11: Comparing Investment Options (Hard)

**Question:**
An investor is considering two stocks, A and B, based on their monthly returns over the last six months:

*   **Stock A Returns:** `[2.5%, 3.0%, 2.8%, 2.7%, 3.2%, 2.9%]`
*   **Stock B Returns:** `[1.0%, 5.0%, -0.5%, 6.0%, 0.5%, 4.0%]`

Which stock would you recommend for an investor looking for consistent, moderate returns, and why? Justify your answer using statistical measures.

**Answer:**

To recommend, we need to calculate the mean and standard deviation for each stock.

**Stock A:**

*   Mean (Average Return) = (2.5 + 3.0 + 2.8 + 2.7 + 3.2 + 2.9) / 6 = **2.85%**
*   Standard Deviation ≈ **0.25%** (low variability)

**Stock B:**

*   Mean (Average Return) = (1.0 + 5.0 + -0.5 + 6.0 + 0.5 + 4.0) / 6 = **2.67%**
*   Standard Deviation ≈ **2.56%** (high variability)

**Recommendation:**

I would recommend **Stock A** for an investor looking for consistent, moderate returns.

**Justification:**

*   **Stock A** has a slightly higher average return (2.85% vs. 2.67%) than Stock B.
*   More importantly, **Stock A** has a significantly lower standard deviation (0.25% vs. 2.56%). This indicates that Stock A's returns are much more consistent and less volatile compared to Stock B, which experiences wider fluctuations (including negative returns). For an investor prioritizing consistency, lower volatility (lower standard deviation) is preferred.
