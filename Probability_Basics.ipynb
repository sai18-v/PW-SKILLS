## Q1. A die is rolled. What is the probability of getting:

### (a) An even number

**Answer:**
Even numbers = {2, 4, 6} = 3 outcomes

Probability = **3/6 = 1/2 = 0.5**

### (b) A number greater than 4

**Answer:**
Numbers greater than 4 = {5, 6} = 2 outcomes

Probability = **2/6 = 1/3 ≈ 0.333**

## Q2. In a class of 50 students:

* 20 like Mathematics (M)
* 15 like Science (S)
* 5 like both

Find the probability that a student likes Mathematics or Science.

**Answer:**

Using:

**P(M ∪ S) = (M + S − Both) / Total**

= (20 + 15 − 5) / 50

= **30/50 = 3/5 = 0.6**

---

## Q3. A bag has 3 red and 2 blue balls. One red ball is drawn without replacement. What is the probability the next ball is also red?

**Answer:**

Remaining balls:

* Red = 2
* Blue = 2

Probability = **2/4 = 1/2 = 0.5**

## Q4. A school has 60% boys and 40% girls. Which sampling method ensures both groups are represented?

**Answer:**

**Stratified Sampling**

**Reason:** The population is divided into boys and girls (strata), and samples are selected from each group to ensure proper representation.

## Q5. Population mean = 160 cm, Sample mean = 158 cm.

Find the sampling error.

**Answer:**

Sampling Error = Sample Mean − Population Mean

= 158 − 160

= **−2 cm**

Magnitude of sampling error = **2 cm**

---

## Q6. Population mean salary = ₹50,000, σ = ₹5,000, Sample size = 100.

Find the Standard Error of Mean (SEM).

**Answer:**

SEM = σ / √n

= 5000 / √100

= 5000 / 10

= **₹500**

## Q9. A factory produces bulbs with a 2% defective rate. If 5 bulbs are selected randomly, what is the probability that all are non-defective?

**Answer:**

Probability of one bulb being non-defective = 0.98

Probability all 5 are non-defective:

= (0.98)^5

≈ **0.9039 (90.39%)**

## Q10. Differentiate between discrete and continuous random variables with examples.

**Answer:**

| Discrete Random Variable               | Continuous Random Variable           |
| -------------------------------------- | ------------------------------------ |
| Takes countable values                 | Takes any value within a range       |
| Usually counts                         | Usually measurements                 |
| Example: Number of students in a class | Example: Height, Weight, Temperature |

**Examples:**

*   **Discrete:** Number of cars in a parking lot.
*   **Continuous:** Time taken to complete an exam.
