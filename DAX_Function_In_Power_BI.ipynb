## Q1. Create a calculated column `TotalPrice` using `Quantity` and `UnitPrice`.

```DAX
TotalPrice =
'Amazon Store Sales'[Quantity] * 'Amazon Store Sales'[UnitPrice]
```

---

## Q2. Create a calculated column `DiscountedPrice`.

```DAX
DiscountedPrice =
'Amazon Store Sales'[TotalPrice] *
(1 - 'Amazon Store Sales'[Discount] / 100)
```

---

## Q3. Create a measure `Average Order Value (AOV)`.

```DAX
Average Order Value =
DIVIDE(
    SUM('Amazon Store Sales'[SalesAmount]),
    DISTINCTCOUNT('Amazon Store Sales'[OrderID])
)
```

---

## Q4. Create a measure for cumulative Year-to-Date (YTD) Sales based on `OrderDate`.

```DAX
YTD Sales =
TOTALYTD(
    SUM('Amazon Store Sales'[SalesAmount]),
    'Amazon Store Sales'[OrderDate]
)
```

---

## Q5. Create a measure to rank each product according to its Total Sales.

```DAX
Product Sales Rank =
RANKX(
    ALL('Amazon Store Sales'[ProductID]),
    CALCULATE(SUM('Amazon Store Sales'[SalesAmount])),
    ,
    DESC,
    DENSE
)
```

---

## Q6. Create a calculated column classifying orders as `"High Value"` if sales > 2000, else `"Low Value"`.

```DAX
Order Value =
IF(
    'Amazon Store Sales'[SalesAmount] > 2000,
    "High Value",
    "Low Value"
)
```

---

## Q7. Create a measure to calculate total sales from the Online channel only.

```DAX
Online Sales =
CALCULATE(
    SUM('Amazon Store Sales'[SalesAmount]),
    'Amazon Store Sales'[SalesChannel] = "Online"
)
```

---

## Q8. Create a measure `Date Difference` that calculates the number of days between the latest Order Date and today's date.

```DAX
Date Difference =
DATEDIFF(
    MAX('Amazon Store Sales'[OrderDate]),
    TODAY(),
    DAY
)
```
