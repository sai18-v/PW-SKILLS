## Q1. What is the difference between a Basic Filter, Advanced Filter, and a Slicer in Power BI?

**Answer:**

* **Basic Filter:** Filters data by selecting specific values from a field.
* **Advanced Filter:** Applies conditions such as *is greater than, contains, begins with,* etc.
* **Slicer:** A visual filter placed directly on the report canvas that allows users to interactively filter other visuals.

---

## Q4. What is a Visual-level Filter, and how does it differ from Page-level and Report-level Filters?

**Answer:**

* **Visual-level Filter:** Affects only the selected visual.
* **Page-level Filter:** Affects all visuals on the current report page.
* **Report-level Filter:** Affects all pages and visuals throughout the report.

---

## Q6. What is the role of Drill-Down in Power BI? How does it improve data exploration?

**Answer:**
Drill-Down allows users to move from a higher level of a hierarchy to more detailed levels, such as **Year → Quarter → Month → Day**. It helps users explore data at different levels of detail without creating separate visuals.

## Q10. Explain the difference between Drill-Down and Drill-Through using a real-world example.

**Answer:**

**Drill-Down:** Moves deeper within the same visual hierarchy.
Example: **Year → Quarter → Month → Day**.

**Drill-Through:** Navigates to another report page containing detailed information about the selected item.
Example: Select **Laptop** on a sales report and navigate to a Product Details page showing its regional sales.

## Q13. What is the “Keep All Filters” option in Drill-Through? How does it affect user interaction?

**Answer:**
**Keep All Filters** preserves the filters applied on the source page when navigating to the drill-through page.

It allows the drill-through page to retain the user's existing filter context instead of showing only the selected drill-through field.
