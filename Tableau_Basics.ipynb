## Q4. What insight can be gained by comparing Sales and Profit together? Then explain how you would create a chart to analyze Profit by Sub-Category.

**Answer:**
Comparing Sales and Profit shows which products generate high revenue and which actually generate high or low profit. A sub-category may have high sales but low or negative profit.

**Chart:**

* Drag **Sub-Category** → Rows
* Drag **Profit** → Columns
* Keep aggregation as **SUM(Profit)**
* Use a **Bar Chart**

This makes profitable and loss-making sub-categories easy to identify.

## Q7. Explain the business importance of geographical analysis. Then describe how you would create a Map showing Sales by State.

**Answer:**
Geographical analysis helps businesses identify high-performing and low-performing locations, understand regional demand, optimize sales strategies, and make better decisions about resource allocation.

**Steps:**

1. Drag **State** into the view.
2. Tableau automatically recognizes it as a geographical field.
3. Select the **Map** visualization.
4. Drag **Sales** onto **Color**.
5. Use **SUM(Sales)**.
6. The map displays Sales by State.

## Q8. What is the difference between a Worksheet and a Dashboard in Tableau? Why is a dashboard more useful for decision-makers?

**Answer:**

* **Worksheet:** Used to create an individual visualization or analysis.
* **Dashboard:** Combines multiple worksheets and visualizations into one interactive view.

A dashboard is more useful for decision-makers because it presents multiple KPIs and insights in one place, allowing them to quickly understand overall business performance and make decisions.

## Q9. What is Tableau Prep, and why is it important in real-world data projects? Explain how you would remove null values, remove duplicate records, and change incorrect data types.

**Answer:**
Tableau Prep is a data preparation tool used to clean, combine, transform, and prepare data before analysis in Tableau.

### Remove null values:

1. Open the dataset in Tableau Prep.
2. Select the required field.
3. Identify null values.
4. Filter out or remove the null records.

### Remove duplicate records:

1. Add a cleaning step.
2. Select the relevant fields.
3. Identify duplicate records.
4. Remove the duplicate rows.

### Change incorrect data types:

1. Select the column.
2. Open the data-type option.
3. Change it to the correct type, such as **Number, Date, or String**.
4. Verify the converted values.
