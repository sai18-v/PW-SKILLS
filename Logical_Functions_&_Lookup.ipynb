### 1. IF Formula (Pass/Fail)

Formula:

```excel
=IF(B2>=40,"Pass","Fail")
```

Results:

* Riya (45) → Pass
* Siya (72) → Pass
* Raju (55) → Pass
* Rahul (80) → Pass
* Ram (30) → Fail

### 2. Bonus Eligibility (Sales > 40000 AND Attendance ≥ 90)

Formula:

```excel
=IF(AND(B2>40000,C2>=90),"Bonus","No Bonus")
```

Results:

| Employee | Result   |
| -------- | -------- |
| Arjun    | Bonus    |
| Simran   | No Bonus |
| Rohit    | No Bonus |
| Anjali   | No Bonus |

### 3. Nested IF Formula (Grades)

Formula:

```excel
=IF(B2>=90,"A",IF(B2>=75,"B",IF(B2>=50,"C","Fail")))
```

Results:

| Student | Marks | Grade |
| ------- | ----- | ----- |
| Raj     | 85    | B     |
| Neha    | 96    | A     |
| Vivek   | 57    | C     |
| Rani    | 92    | A     |
| Sneha   | 75    | B     |

### 4. Department of Employee ID E103

Employee ID **E103** → **Finance**

Formula:

```excel
=VLOOKUP("E103",A2:C5,3,FALSE)
```

**Lookup Function Used:** `VLOOKUP`
**Why:** Employee ID is in the first column and we need a value from a column to its right.

### 5. Commission Rate for Sales of ₹45,000

Commission Table:

| Sales Amount | Rate |
| ------------ | ---- |
| 0            | 5%   |
| 20000        | 10%  |
| 50000        | 15%  |
| 80000        | 20%  |

Formula:

```excel
=VLOOKUP(45000,A2:B5,2,TRUE)
```

✅ **Commission Rate = 10%**

### 7. Salary of EmpID E105 using INDEX-MATCH

Formula:

```excel
=INDEX(B2:B5,MATCH("E105",A2:A5,0))
```

✅ **Salary = 72,000**
