### 1. What is data cleaning, and why is it important in data analysis?

* What are the potential consequences of analyzing unclean or messy data?
* Explain the common steps involved in cleaning and organizing data.

**Answer:**
Data cleaning is the process of identifying and correcting errors, inconsistencies, duplicate records, missing values, and formatting issues in a dataset. It is important because clean data improves accuracy, reliability, and decision-making.

**Consequences of unclean data:**
* Incorrect analysis results
* Poor business decisions
* Reduced productivity
* Inaccurate reports and forecasts

**Common steps:**
1. Remove duplicates
2. Handle missing values
3. Correct formatting errors
4. Standardize data
5. Validate data accuracy
6. Remove irrelevant records

### 3. Use of TRIM, LEFT, RIGHT, MID, and CONCAT

*   **TRIM:** Removes extra spaces.
*   **LEFT:** Extracts characters from the beginning.
*   **RIGHT:** Extracts characters from the end.
*   **MID:** Extracts characters from the middle.
*   **CONCAT:** Combines text from multiple cells.

**Example:**
*   `TRIM(" John ")` → `John`
*   `LEFT("Excel",2)` → `Ex`
*   `RIGHT("Excel",2)` → `el`
*   `MID("Excel",2,3)` → `xce`
*   `CONCAT("Data","Analysis")` → `DataAnalysis`

**Example:**

### 7. Combining Text Using CONCAT, TEXTJOIN, and &

```excel
=CONCAT("Hello"," ","World")
```

**Result:** Hello World

**b. Combine Apple, Banana, and Cherry with commas**

### 8. Understanding TODAY() and NOW()

#### a. Difference between TODAY() and NOW()

| Function | Returns               |
| -------- | --------------------- |
| TODAY()  | Current date only     |
| NOW()    | Current date and time |

**Examples:**

```excel
=TODAY()
```

Returns: 20-06-2026

```excel
=NOW()
```

Returns: 20-06-2026 10:36 AM

---

#### b. Days left until date in A1

If A1 contains **2025-06-10**:

```excel
=A1-TODAY()
```
