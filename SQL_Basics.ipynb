## SQL Basics - Employees Data

```python
import sqlite3
import pandas as pd

# Create/connect to SQLite database
conn = sqlite3.connect("company_db.db")

print("Database connected successfully!")
```

### Q1 — Create employees table

```python
conn.execute("""
CREATE TABLE IF NOT EXISTS employees (
    employee_id INTEGER PRIMARY KEY,
    first_name VARCHAR(50),
    last_name VARCHAR(50),
    department VARCHAR(50),
    salary INTEGER,
    hire_date DATE
)
""")

conn.commit()

print("Employees table created successfully!")
```

### Q2 — Insert employee data

```python
conn.execute("""
DELETE FROM employees
""")

conn.executemany("""
INSERT INTO employees
(employee_id, first_name, last_name, department, salary, hire_date)
VALUES (?, ?, ?, ?, ?, ?)
""", [
    (101, 'Amit', 'Sharma', 'HR', 50000, '2020-01-15'),
    (102, 'Riya', 'Kapoor', 'Sales', 75000, '2019-03-22'),
    (103, 'Raj', 'Mehta', 'IT', 90000, '2018-07-11'),
    (104, 'Neha', 'Verma', 'IT', 85000, '2021-09-01'),
    (105, 'Arjun', 'Singh', 'Finance', 60000, '2022-02-10')
])

conn.commit()

print("Employee data inserted successfully!")
```

```python
pd.read_sql_query("""
SELECT *
FROM employees
""", conn)
```

### Q3 — Salary: Lowest to Highest

```python
employees_by_salary = pd.read_sql_query("""
SELECT *
FROM employees
ORDER BY salary ASC
""", conn)

employees_by_salary
```

### Q4 — Department A–Z and Salary High–Low

```python
employees_by_department_salary = pd.read_sql_query("""
SELECT *
FROM employees
ORDER BY department ASC, salary DESC
""", conn)

employees_by_department_salary
```

### Q5 — IT employees, newest hire first

```python
it_employees = pd.read_sql_query("""
SELECT *
FROM employees
WHERE department = 'IT'
ORDER BY hire_date DESC
""", conn)

it_employees
```

### Q9. What is the Difference Between a Primary Key and a Foreign Key?

**Primary Key:** A column that uniquely identifies each record in a table. It cannot contain duplicate or NULL values.

**Foreign Key:** A column that references the Primary Key of another table and is used to establish a relationship between two tables.

### Q10. What Are Constraints in SQL and Why Are They Used?

**Constraints** are rules applied to columns in a table to maintain **data accuracy, consistency, and integrity**.

Common constraints include:

* **PRIMARY KEY** – uniquely identifies each record.
* **FOREIGN KEY** – establishes a relationship between tables.
* **NOT NULL** – prevents NULL values.
* **UNIQUE** – prevents duplicate values.
* **CHECK** – ensures values meet a specified condition.
* **DEFAULT** – provides a default value when no value is supplied.
