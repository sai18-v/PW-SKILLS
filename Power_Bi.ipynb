## Q1. What is Power BI and why is it used in businesses?

**Answer:**
Power BI is a business intelligence and data visualization tool developed by Microsoft. It is used to connect to data, analyze it, and create interactive reports and dashboards that help businesses make data-driven decisions.

---

## Q2. Name and explain the three main components of Power BI.

**Answer:**

1. **Power BI Desktop** – Used to connect, transform, model, and visualize data and create reports.
2. **Power BI Service** – A cloud-based platform used to publish, share, collaborate on, and manage reports and dashboards.
3. **Power BI Mobile** – Mobile applications used to view and interact with Power BI reports and dashboards on smartphones and tablets.

---

## Q3. Explain the Power BI workflow.

**Answer:**

**Connect → Transform → Model → Visualize → Publish → Share**

1. Connect Power BI to data sources.
2. Transform and clean data using Power Query.
3. Create relationships and build the data model.
4. Create charts, tables, dashboards, and reports.
5. Publish the report to Power BI Service.
6. Share and collaborate with users.

---

## Q5. Write step-by-step instructions to load the Real_Estate Dataset.

**Answer:**

1. Open **Power BI Desktop**.
2. Select **Get Data**.
3. Select the appropriate source containing the Real_Estate Dataset.
4. Select the dataset/table.
5. Click **Transform Data** to open Power Query.
6. Check and clean the data if required.
7. Verify column names and data types.
8. Click **Close & Apply**.
9. The dataset is loaded into Power BI.

## Q6. Define Data View, Report View, and Model View. Explain the purpose of each view.

**Answer:**

1. **Data View** – Displays the actual data in tables and allows you to inspect and analyze rows and columns.
2. **Report View** – Used to create and design visualizations, charts, tables, and interactive reports.
3. **Model View** – Used to view and manage relationships between tables and build the data model.

## Q8. Split Owner Name to create two new columns as First Name and Last Name.

**Answer:**

1. Open **Power Query Editor**.
2. Select the **Owner Name** column.
3. Go to **Transform → Split Column → By Delimiter**.
4. Select **Space** as the delimiter.
5. Choose **Split at the left-most delimiter**.
6. Rename the resulting columns to **First Name** and **Last Name**.
7. Click **Close & Apply**.
