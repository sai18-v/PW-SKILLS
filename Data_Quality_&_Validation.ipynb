## Q1. Define Data Quality in the context of the ETL pipelines. Why is it more than just data cleaning?

**Answer:**
Data quality refers to the accuracy, completeness, consistency, validity, uniqueness, and timeliness of data used in an ETL pipeline. It is more than data cleaning because it also includes validation, monitoring, integrity checks, and ensuring that data is suitable for its intended business purpose.

---

## Q2. Explain why poor data quality leads to misleading dashboards and incorrect decisions.

**Answer:**
Poor-quality data can contain missing, duplicate, inaccurate, or inconsistent values. When this data is used in dashboards, it produces incorrect metrics and misleading reports, which can result in wrong business decisions.

## Q3. What is duplicate data? Explain three causes in ETL pipelines.

**Answer:**
Duplicate data occurs when the same record or business transaction appears more than once.

Three common causes are:

1. Repeated data extraction or loading.
2. Duplicate records in the source system.
3. Incorrect joins or ETL logic that creates multiple copies of records.

---

## Q4. Differentiate between exact, partial, and fuzzy duplicates.

**Answer:**

* **Exact duplicates:** Records are completely identical in all relevant fields.
* **Partial duplicates:** Records match on some important fields but differ in one or more fields.
* **Fuzzy duplicates:** Records represent the same entity but contain small variations such as spelling, formatting, or abbreviations.

---

## Q5. Why should data validation be performed during transformation rather than after loading?

**Answer:**
Validation during transformation helps detect and correct invalid data before it reaches the target system. This prevents bad data from being loaded and ensures that the target database contains reliable and consistent information.

## Q7. Write an SQL query on `Sales_Transactions` to list all duplicate keys and their counts using the business key `(Customer_ID + Product_ID + Txn_Date + Txn_Amount)`.

**Answer:**

```sql
SELECT
    Customer_ID,
    Product_ID,
    Txn_Date,
    Txn_Amount,
    COUNT(*) AS Duplicate_Count
FROM Sales_Transactions
GROUP BY
    Customer_ID,
    Product_ID,
    Txn_Date,
    Txn_Amount
HAVING COUNT(*) > 1;
```

**Result for the given dataset:**

| Customer_ID | Product_ID | Txn_Date   | Txn_Amount | Duplicate_Count |
| ----------- | ---------- | ---------- | ---------: | --------------: |
| C101        | P11        | 2025-12-01 |       4000 |               3 |

## Q8. Enforcing Referential Integrity

**Answer:**

The violating `Customer_ID` is:

**C105**

It appears in `Sales_Transactions` but does not exist in `Customers_Master`.

**SQL query:**

```sql
SELECT DISTINCT s.Customer_ID
FROM Sales_Transactions s
LEFT JOIN Customers_Master c
    ON s.Customer_ID = c.CustomerID
WHERE c.CustomerID IS NULL;
```

**Result:**

```text
C105
```
