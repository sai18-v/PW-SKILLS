## Advanced Visuals in Tableau — Correct Answers

### Q1. Create a Heatmap to analyze Treatment Cost by Department and Region.

**Answer:**

* **Rows:** Department
* **Columns:** Region
* **Marks:** Square
* **Color:** `SUM(Treatment Cost)`
* **Label:** `SUM(Treatment Cost)` (optional)

This shows which department-region combinations have higher or lower treatment costs.

---

### Q2. Create a Highlight Table showing Satisfaction Score by Department.

**Answer:**

* **Rows:** Department
* **Columns:** Satisfaction Score
* **Marks:** Square
* **Color:** `AVG(Satisfaction Score)`
* **Label:** `AVG(Satisfaction Score)`

Use the color intensity to highlight departments with higher or lower satisfaction.

---

### Q3. Create a Bump Chart to show ranking of Departments based on Treatment Cost over months.

**Answer:**

* **Columns:** Month from `Date`
* **Rows:** `RANK(SUM(Treatment Cost))`
* **Color:** Department
* **Marks:** Line
* **Detail:** Department

Set the ranking to **descending**, so rank 1 represents the department with the highest treatment cost.

A bump chart shows how department rankings change over time.

---

### Q4. Create a Butterfly Chart comparing Male vs Female patient count by Department.

**Answer:**

Create two calculated fields:

**Male Patients**

```text
IF [Gender] = "Male" THEN 1 ELSE 0 END
```

**Female Patients**

```text
IF [Gender] = "Female" THEN 1 ELSE 0 END
```

Then:

* **Rows:** Department
* **Columns:** `SUM(Male Patients)` and `SUM(Female Patients)`
* Use a **dual-axis** or mirrored bar-chart approach.
* Reverse the axis for one gender so Male and Female bars face opposite directions.
* **Color:** Gender

This creates a butterfly/tornado-style comparison.

---

### Q5. Create a Dual Axis Chart to compare Average Age and Treatment Cost by Department.

**Answer:**

* **Columns:** Department
* **Rows:** `AVG(Age)`
* Add `SUM(Treatment Cost)` to Rows.
* Right-click the second measure → **Dual Axis**.
* Use **Bar** for Treatment Cost and **Line/Circle** for Average Age.
* Synchronize axes only if appropriate for the visualization.

This compares department-level average age with treatment cost.

---

### Q6. Create a Sunburst Chart showing Department → Diagnosis hierarchy by Treatment Cost.

**Answer:**
Create a hierarchy:

**Department → Diagnosis**

Use:

* **Level 1:** Department
* **Level 2:** Diagnosis
* **Measure:** `SUM(Treatment Cost)`
* **Size:** Treatment Cost
* **Color:** Department/Diagnosis

If the Tableau version does not provide a native Sunburst chart, use a **Sunburst chart extension or a suitable custom visualization**.

The inner ring represents **Department**, while the outer ring represents **Diagnosis**.

---

### Q7. Create a Dual Axis Chart comparing Treatment Cost and Satisfaction Score by Department.

**Answer:**

* **Columns:** Department
* **Rows:** `SUM(Treatment Cost)`
* Add `AVG(Satisfaction Score)` to Rows.
* Select **Dual Axis**.
* Use **Bar** for Treatment Cost.
* Use **Line/Circle** for Satisfaction Score.
* Keep separate axes because the two measures have very different scales.

This allows treatment cost and satisfaction to be compared for each department.

---

### Q8. Create a Dashboard combining Heatmap, Dual Axis Chart, Sunburst Chart, and Butterfly Chart.

**Answer:**

Create four separate worksheets:

1. **Heatmap:** Department × Region → Treatment Cost
2. **Dual Axis Chart:** Department → Treatment Cost + Satisfaction Score
3. **Sunburst:** Department → Diagnosis → Treatment Cost
4. **Butterfly Chart:** Department → Male vs Female Patient Count

Then:

**Dashboard → New Dashboard → Drag all four worksheets onto the dashboard.**

Add appropriate **titles, legends, filters, and tooltips** so users can interactively analyze treatment cost, satisfaction, diagnoses, and patient demographics.
