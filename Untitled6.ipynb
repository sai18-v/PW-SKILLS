## Q1. Category Performance Analysis

**Answer:**
Use a **Bar Chart**.

* **Rows:** Product Category
* **Columns:** Sales
* **Aggregation:** SUM(Sales)

The category with the **longest bar** is the top-performing category by revenue.

---

## Q2. Sales Trend Analysis

**Answer:**
Use a **Line Chart**.

* **Columns:** Order Date
* **Rows:** Sales
* **Aggregation:** SUM(Sales)
* Set Order Date to **Month**.

The chart shows monthly sales fluctuations and helps identify **peak and slow sales periods**.

---

## Q3. Regional Contribution Analysis

**Answer:**
Use a **Bar Chart**.

* **Rows:** Region
* **Columns:** Sales
* **Aggregation:** SUM(Sales)

The region with the **highest SUM(Sales)** is the highest-contributing region.

---

## Q4. Profitability Investigation

**Answer:**
Use a **Scatter Plot**.

* **Columns:** Sales
* **Rows:** Profit
* **Detail:** Order/Product (depending on available field)

Orders with **high Sales but low or negative Profit** are loss-making or low-profit orders. These should be investigated for excessive discounts, high costs, or low margins.

---

## Q5. Business Growth Evaluation

**Answer:**
Use a **Line Chart** showing cumulative sales.

Create a calculated field:

```text
RUNNING_SUM(SUM([Sales]))
```

Then:

* **Columns:** Order Date
* **Rows:** Running Total of Sales
* Set Order Date to **Month**.

An upward-sloping cumulative-sales line indicates overall business growth.

---

## Q6. Category vs Region Performance

**Answer:**
Use a **Heat Map** or **Highlight Table**.

* **Rows:** Region
* **Columns:** Product Category
* **Color:** SUM(Sales)
* **Text:** SUM(Sales)

This allows strong and weak combinations of **Category + Region** to be identified easily.

## Q7. Product Prioritization

**Answer:**
Perform a multi-dimensional analysis using:

* **Product:** Product/Sub-Category
* **Sales:** SUM(Sales)
* **Profit:** SUM(Profit)
* **Quantity:** SUM(Quantity)

A product with **high Sales + high Profit + high Quantity** should receive the highest priority.

Products with **high Sales but low/negative Profit** should be investigated rather than automatically prioritized.

## Q9. Chart Selection Justification

**Answer:**

* **Bar Chart:** Best for comparing categories and regions.
* **Line Chart:** Best for identifying trends and changes over time.
* **Map:** Best for geographical/regional analysis.
* **Scatter Plot:** Best for examining the relationship between Sales and Profit and identifying outliers.
* **Heat Map:** Best for comparing two dimensions, such as Category and Region.

Different charts are selected because each visualization communicates a particular type of business insight most effectively.
