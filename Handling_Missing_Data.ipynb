## Q1. What are the most common reasons for missing data in ETL pipelines?

**Answer:**
Common reasons include:

* Data entry errors
* Missing information at the source
* System or network failures
* Incorrect data extraction
* Data integration issues between different systems

---

## Q2. Why is blindly deleting rows with missing values considered a bad practice in ETL?

**Answer:**
Blindly deleting rows can result in significant data loss and may remove useful information. It can also introduce bias into the dataset, especially when many records contain missing values. The appropriate method should be selected based on the amount and nature of missing data.

---

## Q3. Explain the difference between Listwise deletion and Column deletion. Also mention one scenario where each is appropriate.

**Answer:**

**Listwise deletion:** Removes the entire row if it contains a missing value.
**Appropriate when:** Only a small number of rows have missing values and the remaining data is sufficient.

**Column deletion:** Removes the entire column containing missing values.
**Appropriate when:** A column has a very high percentage of missing values and provides little useful information.

---

## Q4. Why is median imputation preferred over mean imputation for skewed data such as income?

**Answer:**
Median imputation is preferred because the median is less affected by extreme values and outliers. Income data is often highly skewed, so using the mean can produce an unrealistic replacement value.

---

## Q5. What is forward fill and in what type of dataset is it most useful?

**Answer:**
Forward fill replaces a missing value with the most recent previous non-missing value.

It is most useful for **time-series or sequential data**, where the previous value is a reasonable representation of the missing value.

---

## Q6. Why should flagging missing values be done before imputation in an ETL workflow?

**Answer:**
Flagging missing values before imputation preserves information about which values were originally missing. After imputation, the original missing values can no longer be identified directly.

---

## Q7. Consider a scenario where income is missing for many customers. How can missingness itself provide business insights?

**Answer:**
A high number of missing income values may indicate data-entry problems, privacy concerns, a particular customer segment avoiding disclosure, or problems in the source system. Analyzing the missingness pattern can therefore provide useful business and data-quality insights.
