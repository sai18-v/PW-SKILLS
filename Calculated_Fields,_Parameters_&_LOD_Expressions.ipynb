# Correct Answers — Calculated Fields, Parameters & LOD Expressions

## Q1. What is a Calculated Field and why is it important in business dashboards?

**Answer:**
A **Calculated Field** is a new field created using a formula based on existing fields in Tableau.

It is important because it allows analysts to create **custom metrics, calculations, classifications, and KPIs** without changing the original dataset.

**Example:**

```text
Profit Ratio = SUM([Profit]) / SUM([Sales])
```

---

## Q2. What can’t Parameters change visuals directly in Tableau?

**Answer:**
A **Parameter** is a user-controlled value that can be used in calculations, filters, reference lines, and other Tableau features.

A parameter **cannot directly change a visualization by itself**. It must be connected to a **calculated field, filter, or other Tableau logic** to affect the visual.

**Example:** A parameter can let a user choose between Sales and Profit, but a calculated field is needed to make the chart switch between those measures.

---

## Q3. When should EXCLUDE LOD be used?

**Answer:**
Use an **EXCLUDE LOD expression** when you want to calculate a value while **ignoring a particular dimension** from the current view.

**Example:**

```text
{ EXCLUDE [Product_Name] : AVG([Sales]) }
```

This calculates the average Sales while excluding **Product_Name** from the level of detail.

**Use case:** Compare individual product sales with the average sales of the broader category.

---

## Q4. Difference between FIXED and INCLUDE LOD expressions

**Answer:**

| FIXED LOD                                                                          | INCLUDE LOD                                                     |
| ---------------------------------------------------------------------------------- | --------------------------------------------------------------- |
| Calculates at a **specified dimension**, regardless of the dimensions in the view. | Adds a specified dimension to the current level of detail.      |
| Gives a fixed level of aggregation.                                                | Calculates at a more detailed level and can then be aggregated. |
| Example: `{ FIXED [City] : AVG([Sales]) }`                                         | Example: `{ INCLUDE [Product_Name] : SUM([Sales]) }`            |

**In simple terms:**

* **FIXED** → "Calculate specifically at this level."
* **INCLUDE** → "Include this extra level of detail in the calculation."

# Practical Questions

## Q1. Delivery Performance Classification

**Requirement:** Delivery < 30 minutes → Fast; otherwise → Delayed.

### Calculated Field:

```text
IF [Delivery_Time_Min] < 30 THEN
    "Fast"
ELSE
    "Delayed"
END
```

## Q2. Loss-Making Orders Identification

**Requirement:** Identify orders resulting in a loss.

### Calculated Field:

```text
IF [Profit] < 0 THEN
    "Loss-Making"
ELSE
    "Profitable"
END
```

To display only loss-making orders, place this calculated field on **Filters** and select **Loss-Making**.

## Q3. Top N Products Analysis — Parameter

**Answer:**

Create an **Integer Parameter**:

**Parameter name:** `Top N Products`
**Data type:** Integer
**Current value:** 10

Then create a calculated field:

```text
RANK(SUM([Sales]), 'desc') <= [Top N Products]
```

Place the calculated field on **Filters** and select `True`.

This allows the user to dynamically change the number of top products displayed.

## Q4. City-Level Benchmarking — FIXED LOD

**Requirement:** Calculate average sales per city independently of category filters.

### Calculated Field:

```text
{ FIXED [City] : AVG([Sales]) }
```

This calculates the **average Sales for each City** at the city level.

---

## Q5. Customer/Product Detail Impact — INCLUDE LOD

**Requirement:** Calculate average sales at the Product level and allow the result to be used at a higher level such as Category or Region.

### Calculated Field:

```text
{ INCLUDE [Product_Name] : AVG([Sales]) }
```

This includes **Product_Name** in the calculation even when Product_Name is not displayed in the visualization.

**Correct LOD:** `INCLUDE` because the requirement specifically asks to include **product-level detail** in a higher-level analysis.
