#Q1 Explain Difference Between Relative, Absolute and Mixed Cell Referencing
    Vehicle          Price
    Car              500000
    Bicycle          10500
    Auto-rickshaw    25000

Ans:
*   **Relative Cell Referencing**: This is the default behavior in spreadsheets. When a formula with relative references is copied to another cell, the references in the formula automatically change based on the relative position of the original and new cells. For example, if you have a formula in cell B1 that refers to A1, and you copy it to B2, the formula will automatically adjust to refer to A2.

*   **Absolute Cell Referencing**: In absolute referencing, the cell reference remains constant no matter where the formula is copied. This is achieved by placing a dollar sign (`$`) before both the column letter and the row number (e.g., `$A$1`). When a formula containing `$A$1` is copied, it will always refer back to cell A1.

*   **Mixed Cell Referencing**: This is a combination of relative and absolute referencing, where either the column or the row is fixed, but not both. For example:
    *   `A$1`: The column `A` is relative, but the row `1` is absolute. When copied horizontally, the column will change (e.g., to `B$1`), but the row will remain `1`. When copied vertically, the row will remain `1`.
    *   `$A1`: The column `A` is absolute, but the row `1` is relative. When copied horizontally, the column will remain `A`. When copied vertically, the row will change (e.g., to `$A2`).

# Q3 Using the Data Below, Write a Formula to Calculate Average Sales of Items Priced Above 100 but Less Than 300
      item      price     sales
      item A      90       1000
      item B      150      1200
      item C      250      1500

Ans:

First, identify items priced above 100 but less than 300:
*   **Item A**: Price 90 (does not meet criteria)
*   **Item B**: Price 150 (meets criteria), Sales 1200
*   **Item C**: Price 250 (meets criteria), Sales 1500

**Formula:** `(Sales of Item B + Sales of Item C) / Count of qualifying items`

**Calculation:** `(1200 + 1500) / 2 = 2700 / 2 = 1350`

**Average Sales of Items Priced Above 100 but Less Than 300:** 1350

#Q5 Calculate the Total Sales for each row using a formula


To calculate **Total Sales** for each row:

**Formula:**

```excel
=E2*F2
```

Where:

* **E** = Quantity
* **F** = UnitPrice

### Steps in Google Sheets

1. Add a new column named **TotalSales** (e.g., column H).
2. In cell **H2**, enter:

   ```excel
   =E2*F2
   ```
3. Press **Enter**.
4. Drag the fill handle (small square at the bottom-right of the cell) down to copy the formula for all rows.

### Example

| Quantity | UnitPrice | TotalSales |
| -------- | --------- | ---------- |
| 10       | 50        | 500        |
| 20       | 10        | 200        |
| 15       | 25        | 375        |

This will calculate the sales amount for every order automatically.
