### Q1. What is a null hypothesis (H₀) and why is it important?

The **null hypothesis (H₀)** is a statement that there is **no significant difference, effect, or relationship** in the population.

It is important because hypothesis testing starts by assuming H₀ is true and then determines whether there is enough statistical evidence to **reject H₀**.

---

### Q2. What does the significance level (α) represent?

The **significance level (α)** is the probability of making a **Type I error**, meaning rejecting the null hypothesis when it is actually true.

Common values are:

* α = 0.05 → 5% significance level
* α = 0.01 → 1% significance level

---

### Q3. Differentiate between Type I and Type II errors.

| Error             | Meaning                      |
| ----------------- | ---------------------------- |
| **Type I Error**  | Rejecting a true H₀          |
| **Type II Error** | Failing to reject a false H₀ |

In simple terms:

* **Type I:** False positive
* **Type II:** False negative

---

### Q4. Difference between one-tailed and two-tailed tests

**One-tailed test:** Tests for an effect in **one specific direction**.

Example:

> H₁: The average salary is **greater than** ₹50,000.

**Two-tailed test:** Tests whether there is a difference in **either direction**.

Example:

> H₁: The average salary is **different from** ₹50,000.

### Q5. Company claims average complaint resolution time is 10 minutes

Given:

* μ₀ = 10 minutes
* Sample mean = 12 minutes
* s = 3 minutes
* n = 9
* α = 0.05

Since **sample size is small and population standard deviation is unknown**, use a **one-sample t-test**.

**Hypotheses:**

H₀: μ = 10
H₁: μ ≠ 10

Test statistic:

[ t=\frac{\bar{x}-\mu_0}{s/\sqrt n} ]

[ t=\frac{12-10}{3/\sqrt9} ]

[ t=\frac{2}{1}=2 ]

Degrees of freedom:

[ df=9-1=8 ]

At α = 0.05 for a two-tailed test:

[ t_{critical}=±2.306 ]

Since:

[ |2| < 2.306 ]

**Decision:** Fail to reject H₀.

**Conclusion:** There is **not enough evidence at the 5% significance level to reject the company's claim** that the average resolution time is 10 minutes.

### Q6. When should you use a z-test instead of a t-test?

Use a **z-test** when:

* The population standard deviation **σ is known**, and
* The population is normally distributed or the sample size is sufficiently large.

A **t-test** is generally used when the population standard deviation is **unknown**, especially for small samples.

### Q7. Productivity before and after training

Data:

| Employee | Before | After | Difference |
| -------- | -----: | ----: | ---------: |
| 1        |     50 |    55 |          5 |
| 2        |     60 |    65 |          5 |
| 3        |     58 |    59 |          1 |
| 4        |     55 |    58 |          3 |
| 5        |     62 |    63 |          1 |
| 6        |     56 |    59 |          3 |

Because the **same employees are measured before and after**, use a **paired t-test**.

**Hypotheses:**

H₀: μd = 0
H₁: μd > 0

Mean difference:

[ \bar d = \frac{5+5+1+3+1+3}{6}=3 ]

Standard deviation of differences:

[ s_d \approx 1.789 ]

Test statistic:

[ t=\frac{\bar d}{s_d/\sqrt n} ]

[ t=\frac{3}{1.789/\sqrt6} ]

[ t\approx4.108 ]

Degrees of freedom:

[ df=6-1=5 ]

For a one-tailed test at α = 0.05:

[ t_{critical}\approx2.015 ]

Since:

[ 4.108 > 2.015 ]

**Decision:** Reject H₀.

**Conclusion:** There is **significant evidence that the training program improved employee productivity**.

### Q8. Test whether product preference is independent of gender

Observed table:

| Gender | Product A | Product B | Total |
| ------ | --------: | --------: | -----:|
| Male   |        30 |        20 |    50 |
| Female |        10 |        40 |    50 |
| Total  |        40 |        60 |   100 |

This requires a **Chi-square test of independence**.

**Hypotheses:**

H₀: Product preference is **independent of gender**.

H₁: Product preference is **not independent of gender**.

Expected frequencies:

| Gender | Product A | Product B |
| ------ | --------: | --------: |
| Male   |        20 |        30 |
| Female |        20 |        30 |

Chi-square statistic:

[ \chi^2=\sum\frac{(O-E)^2}{E} ]

[ \chi^2 =
\frac{(30-20)^2}{20}
+\frac{(20-30)^2}{30}
+\frac{(10-20)^2}{20}
+\frac{(40-30)^2}{30}
]

[ \chi^2 \approx16.667 ]

Degrees of freedom:

[ df=(2-1)(2-1)=1 ]

At α = 0.05:

[ \chi^2_{critical}=3.841 ]

Since:

[ 16.667 > 3.841 ]

**Decision:** Reject H₀.
