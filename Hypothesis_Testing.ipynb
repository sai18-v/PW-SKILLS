## Correlation and Covariance

### Q1. Define covariance and explain how it differs from correlation in terms of scale and interpretation.

**Covariance** measures the direction in which two variables change together.

* Positive covariance → both variables tend to increase/decrease together.
* Negative covariance → one variable tends to increase while the other decreases.
* Zero covariance → no linear relationship is indicated.

**Difference from correlation:**

* Covariance depends on the **units/scales** of the variables.
* Correlation is **standardized** and always lies between **−1 and +1**.
* Correlation is therefore easier to interpret and compare across datasets.

---

### Q2. What do positive, negative, and zero covariance indicate?

* **Positive covariance:** Both variables tend to move in the same direction.
* **Negative covariance:** Variables tend to move in opposite directions.
* **Zero covariance:** No linear relationship between the variables is indicated.

---

### Q3. Limitations of covariance. Why is correlation preferred?

Covariance has the following limitations:

1. Its value depends on the **units of measurement**.
2. It is difficult to interpret the **strength** of the relationship from covariance alone.
3. Covariance values cannot easily be compared between different datasets.

**Correlation is preferred** because it is standardized between **−1 and +1**, making both the direction and strength of a linear relationship easier to interpret.

---

### Q4. Difference between Pearson's correlation and Spearman's rank correlation

**Pearson correlation:**

* Measures the **linear relationship** between two continuous variables.
* Uses the actual numerical values.
* Sensitive to outliers.

**Spearman's rank correlation:**

* Measures the relationship between the **ranks** of variables.
* Suitable for ordinal data or non-normally distributed data.
* Measures a **monotonic relationship**.
* Less sensitive to outliers.

**Use Spearman's correlation** when the data are ordinal/ranked, not normally distributed, or when the relationship is monotonic but not necessarily linear.

---

### Q5. Correlation coefficient = 0.85

A correlation coefficient of:

[ r=0.85 ]

indicates a **strong positive correlation** between X and Y.

This means that as X increases, Y generally tends to increase as well.

**Can we infer causation? — No.**

Correlation does **not** imply causation. A third variable or other factors could be responsible for the observed relationship.

### Q6. Calculate covariance between X and Y

Given:

| X |  2 |  4 |  6 |  8 |
| - | -: | -: | -: | -: |
| Y |  3 |  7 |  5 | 10 |

Means:

[ \bar X=\frac{2+4+6+8}{4}=5 ]

[ \bar Y=\frac{3+7+5+10}{4}=6.25 ]

Calculate:

[ \sum(X-\bar X)(Y-\bar Y)=19 ]

Using **sample covariance**:

[ Cov(X,Y)=\frac{19}{4-1} ]

[ \boxed{Cov(X,Y)=6.33} ]

So, the covariance is **approximately 6.33**, indicating a **positive relationship**.

> If your course uses the population covariance formula, the answer would be **4.75**. For statistical sample data, **6.33** is normally reported.

### Q7. Pearson correlation coefficient between A and B

Given:

| A | 10 | 20 | 30 | 40 | 50 |
| - | -: | -: | -: | -: | -: |
| B |  8 | 14 | 18 | 24 | 28 |

Using Pearson's correlation formula:

[ r=\frac{n\sum AB-\sum A\sum B}
{\sqrt{[n\sum A^2-(\sum A)^2][n\sum B^2-(\sum B)^2]}} ]

The result is:

[ \boxed{r\approx0.994} ]

Therefore, A and B have a **very strong positive correlation**.

# Q10. Portfolio A and Portfolio B

Given:

| Year | Portfolio A | Portfolio B |
| ---- | ----------: | ----------: |
| 1    |           8 |           6 |
| 2    |          10 |           9 |
| 3    |          12 |          11 |
| 4    |           9 |           8 |
| 5    |          11 |          10 |

### Step 1: Covariance

Mean of A:

[ \bar A=\frac{8+10+12+9+11}{5}=10 ]

Mean of B:

[ \bar B=\frac{6+9+11+8+10}{5}=8.8 ]

The sum of cross-deviations is:

[ \sum(A-\bar A)(B-\bar B)=12 ]

Using sample covariance:

[ Cov(A,B)=\frac{12}{5-1} ]

[ \boxed{Cov(A,B)=3} ]

### Step 2: Correlation coefficient

[ \boxed{r\approx0.986} ]

### Interpretation

The correlation is very close to **+1**, so the two portfolios have a **very strong positive correlation**.

Therefore, they generally **move together**.
